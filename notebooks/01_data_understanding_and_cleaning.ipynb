# Week 1 — Food Delivery Dataset Understanding & Cleaning

**Project:** Kagnus Food Delivery Operations  
**Objective:** Inspect the supplied dataset, identify and document quality issues, make conservative justified corrections, and verify a reproducible cleaned CSV.  
**Dataset source:** Kagnus project brief (link in repository README).  
**Unit of observation:** One row appears to represent one delivery record, supported by the unique `ID`, order details, delivery-partner fields and `Time_taken (min)`. This interpretation is based on column evidence, not an external data dictionary.

**Run instructions:** Put the approved original CSV at `data/raw/Food delivery Dataset.csv` and run this notebook from the repository root. No records are dropped merely for missing values or statistical outliers.

## 1. Load and inspect the dataset
We load the raw CSV without modifying it, then inspect the first and last records, dimensions, names, data types and summary statistics. This establishes what is actually available.

```python
from pathlib import Path
import pandas as pd
import numpy as np
from IPython.display import display, Markdown

# Works in a local repository, or from the notebooks folder.
root = Path.cwd()
if not (root / 'data').exists() and (root.parent / 'data').exists():
    root = root.parent
raw_path = root / 'data/raw/Food delivery Dataset.csv'
if not raw_path.exists():
    raise FileNotFoundError(f'Place the Kagnus-approved CSV at {raw_path}')
df = pd.read_csv(raw_path)
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.head(), df.tail())
df.info()
display(df.describe(include='all').T)
```

### Column overview
We describe each observed column conservatively. Latitude and longitude are coordinates; `Time_taken (min)` is in minutes by its label. Other fields with undocumented units or codes are not assigned invented meanings.

```python
meanings = {
'ID':'Delivery-record identifier', 'Delivery_person_ID':'Delivery partner identifier',
'Delivery_person_Age':'Reported delivery partner age (years, assumed)',
'Delivery_person_Ratings':'Reported delivery partner rating (scale not documented)',
'Restaurant_latitude':'Restaurant latitude (degrees)', 'Restaurant_longitude':'Restaurant longitude (degrees)',
'Delivery_location_latitude':'Delivery latitude (degrees)', 'Delivery_location_longitude':'Delivery longitude (degrees)',
'Order_Date':'Order date (day-month-year format observed)', 'Time_Orderd':'Order time (HH:MM observed; source spelling)',
'Time_Order_picked':'Pickup time (HH:MM observed)', 'Weather_conditions':'Reported weather category',
'Road_traffic_density':'Reported traffic category', 'Vehicle_condition':'Vehicle condition code (meaning undocumented)',
'Type_of_order':'Order category', 'Type_of_vehicle':'Vehicle category',
'multiple_deliveries':'Reported number/code of multiple deliveries (definition undocumented)',
'Festival':'Festival indicator', 'City':'City classification, not city name',
'Time_taken (min)':'Delivery duration (minutes)'}
column_overview = pd.DataFrame({'column':df.columns,'dtype':[str(x) for x in df.dtypes],
'meaning':[meanings.get(c,'Not documented') for c in df.columns]})
display(column_overview)
```

## 2. Data-quality checks
Count missing values, full-row duplicates and duplicate record IDs. Then inspect categorical labels, numeric ranges and parseability of dates/times. We investigate rather than delete unusual values automatically.

```python
missing = pd.DataFrame({'column':df.columns, 'missing_count':df.isna().sum().values,
'missing_pct':(df.isna().mean().values*100).round(2)})
full_duplicates = int(df.duplicated().sum())
id_duplicates = int(df['ID'].duplicated().sum()) if 'ID' in df.columns else None
print('Exact duplicate rows:', full_duplicates, '| Duplicate IDs:', id_duplicates)
display(missing)
category_cols = ['Weather_conditions','Road_traffic_density','Type_of_order','Type_of_vehicle','Festival','City']
for col in category_cols:
    if col in df:
        print(f'\n{col}:', df[col].value_counts(dropna=False).to_dict())
num_summary = df.select_dtypes(include='number').agg(['min','max']).T
num_summary['negative_count'] = [(df[c]<0).sum() for c in num_summary.index]
display(num_summary)
# Coordinates may be negative legitimately; inspect rather than treating every negative as invalid.
print('Ratings above 5 (potentially invalid if scale is 1–5):', int((df['Delivery_person_Ratings']>5).sum()))
print('Nonpositive duration:', int((df['Time_taken (min)']<=0).sum()))
print('Restaurant coordinate sign mismatches:',int(((df.Restaurant_latitude<0)|(df.Restaurant_longitude<0)).sum()))
print('Potential placeholder delivery coordinates:',int(((df.Delivery_location_latitude==0.01)|(df.Delivery_location_longitude==0.01)).sum()))
parsed_dates = pd.to_datetime(df['Order_Date'], format='%d-%m-%Y', errors='coerce')
print('Unparseable order dates:',int((parsed_dates.isna() & df['Order_Date'].notna()).sum()))
for c in ['Time_Orderd','Time_Order_picked']:
    parsed = pd.to_datetime(df[c].astype('string'), format='%H:%M', errors='coerce')
    print(c, 'unparseable nonmissing:',int((parsed.isna() & df[c].notna()).sum()))
```

### Quality-check interpretation
The tables above are calculated from the provided CSV. Missing values must not be replaced with made-up observations. `ID` identifies delivery records; repeated `Delivery_person_ID` values are expected because one partner can handle many deliveries. Negative restaurant coordinates and unusually high ratings warrant source validation; we retain them rather than guessing corrections. No statistical outlier is removed automatically.

## 3. Conservative cleaning and cleaning log
We copy `df` to `cleaned_df`. We trim surrounding whitespace from text labels and standardize only identical labels differing by whitespace/case. We preserve missing values, uncertain ratings and suspicious coordinates for review. Every investigated issue is logged with actual affected-row counts.

```python
cleaned_df = df.copy(deep=True)
log = []
def record(issue, col, count, action, reason):
    log.append({'issue':issue,'affected_column':col,'affected_rows':int(count),
                'action_taken':action,'reason':reason})
for col in category_cols:
    if col not in cleaned_df: continue
    original = cleaned_df[col].astype('string')
    stripped = original.str.strip()
    # Only harmonize whitespace and capitalization; do not guess spelling corrections.
    canonical = {v.casefold():v for v in sorted(stripped.dropna().unique(), key=lambda s:(s.casefold(),s))}
    standardized = stripped.map(lambda v: canonical.get(v.casefold(),v) if pd.notna(v) else pd.NA)
    changed = int((original.fillna('<MISSING>') != standardized.fillna('<MISSING>')).sum())
    cleaned_df[col] = standardized.astype('object').where(standardized.notna(), np.nan)
    record('Whitespace/case label consistency',col,changed,
           'Trimmed and harmonized labels' if changed else 'Checked; no changes',
           'Only equivalent labels were standardized; no unsupported spelling assumptions')
for col in df.columns:
    count = int(df[col].isna().sum())
    if count:
        record('Missing values',col,count,'Retained as missing',
               'No validated source or defensible imputation rule is provided')
record('Exact duplicate records','all columns',full_duplicates,'Retained; none removed',
       'Duplicate check completed; no automatic deletion')
if id_duplicates is not None:
    record('Duplicate record IDs','ID',id_duplicates,'Retained; none removed',
           'Record ID uniqueness checked')
record('Ratings above 5','Delivery_person_Ratings',int((df['Delivery_person_Ratings']>5).sum()),
       'Flagged; retained','Rating scale not confirmed by source; cannot safely correct')
record('Negative restaurant coordinates','Restaurant_latitude / Restaurant_longitude',
       int(((df.Restaurant_latitude<0)|(df.Restaurant_longitude<0)).sum()),
       'Flagged; retained','Negative coordinates are not intrinsically invalid; requires geospatial validation')
record('Potential placeholder delivery coordinates','Delivery_location_latitude / Delivery_location_longitude',
       int(((df.Delivery_location_latitude==0.01)|(df.Delivery_location_longitude==0.01)).sum()),
       'Flagged; retained','Coordinates need external verification before modification')
record('Nonpositive delivery duration','Time_taken (min)',int((df['Time_taken (min)']<=0).sum()),
       'Checked; retained','Duration range checked; no automatic record removal')
record('Order date parseability','Order_Date',int((parsed_dates.isna() & df['Order_Date'].notna()).sum()),
       'Validated; original text preserved','Day-month-year format verified without altering source values')
for col in ['Time_Orderd','Time_Order_picked']:
    parsed = pd.to_datetime(df[col].astype('string'), format='%H:%M', errors='coerce')
    record('Time parseability',col,int((parsed.isna() & df[col].notna()).sum()),
           'Validated; original text preserved','HH:MM format checked; no timezone or date invented')
cleaning_log = pd.DataFrame(log,columns=['issue','affected_column','affected_rows','action_taken','reason'])
display(cleaning_log)
```

## 4. Export and verify
Save the cleaned data separately and reopen the exported CSV. Compare original and cleaned row counts, column counts and the record IDs to ensure no records were unintentionally removed.

```python
processed = root/'data/processed/cleaned_food_delivery.csv'
log_path = root/'docs/cleaning_log.csv'
processed.parent.mkdir(parents=True,exist_ok=True)
log_path.parent.mkdir(parents=True,exist_ok=True)
cleaned_df.to_csv(processed,index=False)
cleaning_log.to_csv(log_path,index=False)
reloaded = pd.read_csv(processed)
comparison = pd.DataFrame({'metric':['Rows','Columns','Exact duplicate rows'],
'original':[len(df),len(df.columns),int(df.duplicated().sum())],
'cleaned':[len(reloaded),len(reloaded.columns),int(reloaded.duplicated().sum())]})
display(comparison)
assert len(reloaded)==len(df), 'Unexpected change in record count'
assert reloaded.columns.tolist()==df.columns.tolist(), 'Unexpected column changes'
assert reloaded['ID'].astype(str).tolist()==df['ID'].astype(str).tolist(), 'Record IDs changed'
print('CSV export and reload verified:',processed)
print('Cleaning log saved:',log_path)
print('Rows removed: 0; uncertain anomalies and missing values intentionally retained.')
```

## 5. Results, decisions and limitations

- **Record count:** The executed before/after table verifies that no delivery records were removed. This is intentional: the available evidence does not justify dropping missing records or anomalies.
- **Missing values:** The executed quality table gives actual counts and percentages. Missing values remain missing because inventing age, weather, city or ratings would distort analysis.
- **Duplicate checks:** Exact duplicate rows and duplicate `ID` values were checked separately. Repeated delivery-person identifiers are not treated as duplicates.
- **Categorical values:** Whitespace/capitalization checks are applied to the six observed categorical columns. See the cleaning log for actual changes (including zero changes).
- **Numeric anomalies:** Ratings above five and unusual coordinates are flagged, not silently changed; the rating scale and geographic validity need confirmation.
- **Dates and units:** Order dates are tested as DD-MM-YYYY and times as HH:MM. The duration field explicitly indicates minutes. Other coded-field meanings are not assumed.
- **Limitations:** There is no authoritative column dictionary, validated geographic reference or confirmed rating scale. The cleaned file is suitable for cautious next-stage analysis, with flagged anomalies requiring follow-up. No causal conclusions are drawn.
